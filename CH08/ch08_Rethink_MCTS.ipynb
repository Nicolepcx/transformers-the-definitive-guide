# About this notebook

This notebook demonstrates how to clone and run the **RethinkMCTS** project, an implementation of a **Monte Carlo Tree Search (MCTS)-based reasoning algorithm** designed for evaluating and improving large language models on program synthesis tasks such as **HumanEval**.

## What you will run

1. Clone the RethinkMCTS GitHub repository.
2. Change into the repo directory (`%cd RethinkMCTS`).
3. Install the required Python libraries.
4. Launch `run.py` with specific parameters:

   * `--rollout 16`: number of rollouts per MCTS step
   * `--start 4000 --end 4001`: dataset slice to evaluate (here, one HumanEval problem)
   * `--arch gpt4o-mini`: model backbone used for generation
   * `--model rethinkmcts`: selects the RethinkMCTS algorithm variant
   * `--dataset humaneval`: dataset for evaluation

## How it works

* **MCTS search**: RethinkMCTS expands multiple candidate reasoning paths via rollouts, scores them, and backpropagates results to balance exploration and exploitation.
* **Rollouts (`--rollout`)**: Control how many reasoning trajectories are simulated per decision. More rollouts improve search quality at higher compute cost.
* **Dataset slice (`--start`, `--end`)**: Lets you run on a subset of tasks for quick testing or scaling up to full runs.
* **Backbone model (`--arch`)**: Specifies which LLM is used to generate candidate solutions.
* **Algorithm choice (`--model`)**: Allows comparison of RethinkMCTS against baseline strategies.
* **Evaluation**: Integrates with HumanEval, where correctness is judged by executing generated code against test cases.

## Why this pattern

* Reproducible: By pinning dataset slices, you can rerun the same experiment deterministically.
* Modular: Easy to swap backbone models or adjust rollout budgets.
* Scalable: Small settings for debugging, larger ones for full benchmark evaluation.
* Standardized: HumanEval provides a consistent way to compare reasoning/search approaches for program synthesis.

## Extend and adapt

* Increase `--rollout` for deeper exploration.
* Run larger slices (`--start`, `--end`) to cover more HumanEval tasks.
* Try different `--arch` models (e.g., other GPT variants or open-source LLMs).
* Add new datasets or problem sets beyond HumanEval.
* Modify scoring functions in `run.py` to test alternative evaluation criteria.

## Requirements and notes

* Requires a working PyTorch installation with GPU support for efficiency.
* Ensure your environment has access to an appropriate LLM (local or API-backed).
* Rollout counts and long token limits may require substantial compute resources.
* Execution correctness is tied to deterministic test cases in HumanEval.
