# About this notebook

This notebook demonstrates how to call Kimi K2 **via Novita AI’s OpenAI-compatible API**. It shows both a minimal “hello world” chat run and a more advanced configuration with full decoding parameters for code generation tasks.

## What it shows

* **Client setup** with `OpenAI` using a custom `base_url` (`https://api.novita.ai/v3/openai`) and your Novita API key.
* Kimi K2 Instruct** (`moonshotai/kimi-k2-instruct`) with detailed decoding controls
* **Streaming vs non-streaming output** printing.
* **Decoding knobs**: `temperature`, `top_p`, `min_p`, `top_k`, presence/frequency penalties, and repetition penalty.
* **System vs user messages**: system role defines assistant behavior, user role provides the task.

## Extend and adapt

* Make sure your environment supports streaming when `stream=True`.

# Setup model and parameters

```python
model = model = "moonshotai/kimi-k2-instruct"
stream = True # or False
max_tokens = 65536
system_content = "You are a helpful coding assistant"
temperature = 0.7
top_p = 0.8
min_p = 0
top_k = 20
presence_penalty = 0
frequency_penalty = 0
repetition_penalty = 1.05
response_format = { "type": "text" }
```

# Run inference

```python


chat_completion_res = client.chat.completions.create(
    model=model,
    messages=[
        {
            "role": "system",
            "content": system_content,
        },
        {
            "role": "user",
            "content": "Write a quick sort algorithm.",
        }
    ],
    stream=stream,
    max_tokens=max_tokens,
    temperature=temperature,
    top_p=top_p,
    presence_penalty=presence_penalty,
    frequency_penalty=frequency_penalty,
    response_format=response_format,
    extra_body={
      "top_k": top_k,
      "repetition_penalty": repetition_penalty,
      "min_p": min_p
    }
  )

if stream:
    for chunk in chat_completion_res:
        print(chunk.choices[0].delta.content or "", end="")
else:
    print(chat_completion_res.choices[0].message.content)

```
